<h2>🧹 Cleaning Categorical Text Data</h2>

<p style="font-size:15px; line-height:1.8; text-align:justify;">

In this step, I clean the categorical text columns to improve data consistency and ensure accurate analysis. Text cleaning is an important preprocessing step that helps eliminate formatting inconsistencies which may otherwise create duplicate categories or affect analytical results.

</p>

<p style="font-size:15px; line-height:1.8; text-align:justify;">

The following cleaning operations are performed:

</p>

<ul style="font-size:15px; line-height:1.8;">
    <li>Remove leading and trailing spaces using the <b>strip()</b> function.</li>
    <li>Convert all text values to lowercase using the <b>lower()</b> function to maintain consistent formatting.</li>
    <li>Remove unnecessary punctuation, such as the full stop (<b>.</b>) from values like <b>admin.</b>, resulting in <b>admin</b>.</li>
</ul>

<p style="font-size:15px; line-height:1.8; text-align:justify;">

Standardizing the categorical data ensures that similar values are treated as a single category, improves data quality, and enhances the accuracy of exploratory data analysis (EDA), statistical summaries, visualizations, and predictive modeling.

</p>

</div>

```python
# List of categorical text columns
text_cols = [
    'job',
    'marital_status',
    'education',
    'credit_in_default',
    'housing',
    'loan',
    'contact',
    'month',
    'day_of_week',
    'previous_outcome',
    'subscription_status'
]

# Clean text columns
for col in text_cols:
    df[col] = (
        df[col]
        .str.strip()
        .str.lower()
        .str.replace('.', '', regex=False)  # Removes '.' from values like 'admin.'
    )
```

<h2>👥 Creating Age Groups</h2>

<p style="font-size:15px; line-height:1.8; text-align:justify;">

In this step, I group the <b>age</b> column into meaningful categories using the <b>pd.cut()</b> function. Rather than analyzing individual ages, categorizing customers into age groups makes the data easier to interpret and compare across different segments.

</p>

<p style="font-size:15px; line-height:1.8; text-align:justify;">

The following age groups are created:

</p>

<ul style="font-size:15px; line-height:1.8;">
    <li><b>Young</b> : 18–30 years</li>
    <li><b>Adult</b> : 31–45 years</li>
    <li><b>Middle Age</b> : 46–60 years</li>
    <li><b>Senior</b> : 61–100 years</li>
</ul>

<p style="font-size:15px; line-height:1.8; text-align:justify;">

Creating these age groups simplifies customer segmentation and enables more meaningful analysis of customer behavior, campaign effectiveness, and subscription patterns across different age categories. It also supports clearer visualizations and comparative analysis during exploratory data analysis (EDA).

</p>

</div>

```python
bins = [18,30,45,60,100]

labels = ['young','adult','middle age','senior']

df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels)
```
